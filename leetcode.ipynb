Best Time to Buy and Sell Stock (LeetCode 121, easy)
You get an array prices where prices[i] is a stock’s price on day i. You buy on one day and sell on a later day. Return the biggest profit you can make, or 0 if no trade makes money.
Example: prices = [7,1,5,3,6,4] → 5 (buy at 1, sell at 6).


```python
def trade(prices):
  lowest = float('inf')
  best = 0
  for i in prices :
    lowest = min (lowest,i)
    best = max(best,i-lowest)
  return best

print(trade([7,6,5,4,3,2]))
```

A biker starts a trip at altitude 0. You’re given an array gain, where gain[i] is the net change in altitude between point i and point i+1. Return the highest altitude reached at any point.

Example: gain = [-5, 1, 5, 0, -7] → 1 (the altitudes are 0, -5, -4, 1, 1, -6)

```python
def max_altitude(array):
  L=[0]
  for i in range (1,len(array)+1):
    L.append(L[i-1]+array[i-1])
  return max(L)
print (max_altitude([-5, 1, 5, 0, -7]))
```

```python
def max_altitude(array):
  altitude=0
  best_altitude=0
  for gain in array :
    altitude+=gain
    best_altitude=max(altitude,best_altitude)
  return best_altitude
print (max_altitude([-5, 1, 5, 0, -7]))
```

Final Prices With a Special Discount in a Shop (LeetCode 1475, easy)

You’re given prices, where prices[i] is the price of item i. If you buy item i, you get a discount equal to prices[j], where j is the first index after i with prices[j] <= prices[i]. If there’s no such j, you get no discount. Return the final price of every item.

Example: prices = [8, 4, 6, 2, 3] → [4, 2, 4, 2, 3]

```python
def new_price(prices):
  new_prices=[]
  for i in range(0,len(prices)-1) :
        a=1
        while i + a < len(prices) and prices[i] < prices[i + a]:
          a += 1

        if i + a < len(prices):
          new_prices.append(prices[i] - prices[i + a])
        else:
          new_prices.append(prices[i])

  new_prices.append(prices[-1])
  return new_prices
print(new_price([8, 4, 6, 2, 3]))
```
